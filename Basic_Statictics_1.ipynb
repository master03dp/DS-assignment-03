## Descriptive Analytics and Data Preprocessing on Sales & Discounts Dataset

### Introduction
To begin, we'll load the dataset and perform an initial inspection to understand its structure and content.

```python
import pandas as pd

# Load the dataset
try:
    df = pd.read_csv('sales_data_with_discounts (1).csv')
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: 'sales_data_with_discounts (1).csv' not found. Please ensure the correct file is uploaded and named appropriately.")
    df = pd.DataFrame() # Create an empty DataFrame to prevent errors in subsequent cells if file is missing
```

### Dataset Overview
Let's display the first 5 rows of the dataset and check its general information (data types, non-null values, etc.).

```python
# Display the first 5 rows of the DataFrame
display(df.head())

# Display basic information about the DataFrame
print(df.info())
```

### Descriptive Analytics for Numerical Columns
We will now identify numerical columns and compute their basic statistical measures (mean, median, mode, and standard deviation).

```python
# Identify numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns

if not numerical_cols.empty:
    print(f"Identified numerical columns: {list(numerical_cols)}")

    # Calculate descriptive statistics
    # Mean
    mean_stats = df[numerical_cols].mean()
    # Median
    median_stats = df[numerical_cols].median()
    # Mode (can return multiple values, so we take the first)
    mode_stats = df[numerical_cols].mode().iloc[0] if not df[numerical_cols].empty else pd.Series(dtype='float64')
    # Standard Deviation
    std_stats = df[numerical_cols].std()

    # Combine into a single DataFrame for better presentation
    descriptive_stats = pd.DataFrame({
        'Mean': mean_stats,
        'Median': median_stats,
        'Mode': mode_stats,
        'Standard Deviation': std_stats
    })

    print("\nDescriptive Statistics for Numerical Columns:")
    display(descriptive_stats)
else:
    print("No numerical columns found in the dataset.")
```

### Interpretation of Numerical Statistics
Based on the descriptive statistics above, we can gain initial insights into the central tendency, spread, and common values for each numerical column.

### Data Visualization
We will now visualize the distribution of numerical columns using histograms.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for plots
sns.set_style("whitegrid")

# Plot histograms for numerical columns
print("Generating histograms for numerical columns...")

if not numerical_cols.empty:
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(numerical_cols):
        plt.subplot(len(numerical_cols)//3 + 1, 3, i + 1) # Adjust subplot grid dynamically
        sns.histplot(df[col], kde=True)
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()
    print("Histograms generated successfully.")
else:
    print("No numerical columns to plot histograms for.")
```

### Box Plots
Next, we will generate box plots for numerical variables to visualize their distribution, identify potential outliers, and understand their spread.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot box plots for numerical columns
print("Generating box plots for numerical columns...")

if not numerical_cols.empty:
    plt.figure(figsize=(15, 10))
    for i, col in enumerate(numerical_cols):
        plt.subplot(len(numerical_cols)//3 + 1, 3, i + 1) # Adjust subplot grid dynamically
        sns.boxplot(y=df[col])
        plt.title(f'Box Plot of {col}')
        plt.ylabel(col)
    plt.tight_layout()
    plt.show()
    print("Box plots generated successfully.")
else:
    print("No numerical columns to plot box plots for.")
```
